Task 1: Temperatures.

```python
%%writefile temperature.py
from functools import total_ordering
from math import isclose

@total_ordering
class Temperature:
    """Represent a temperature in Celsius, Fahrenheit, or Kelvin."""

    _TOLERANCE = 1e-9
    _ABSOLUTE_ZERO = {"C": -273.15, "F": -459.67, "K": 0.0}

    def __init__(self, value: float, unit: str = "C") -> None:
        """Create  a temperature, raise ValueError for invalid unit og value"""
        self._unit = self._validate_unit(unit)
        self._value = float(value)

        if self._value < self._ABSOLUTE_ZERO[self._unit]:
            raise ValueError("Temperature cannot be below absolute zero.")


    @staticmethod
    def _validate_unit(unit: str) -> str:
        """Return the uppercase unit or ValueError if the unit is invalid."""
        if not isinstance(unit, str) or unit.upper() not in ("C", "F", "K"):
            raise ValueError("Unit must be 'C', 'F' or 'K'.")
        return unit.upper()


    @property
    def value(self) -> float:
        """Return the temperature value in its original unit."""
        return self._value


    @property
    def unit(self) -> str:
        """Return the uppercase unit symbol."""
        return self._unit


    def to_celsius(self) -> float:
        """Return the temperature in Celsius."""
        if self.unit == "F":
            return (self.value-32) * 5/9
        if self.unit == "K":
            return self.value - 273.15
        return self.value


    def to_fahrenheit(self) -> float:
        """Return the temperature in Fahrenheit."""
        if self.unit == "F":
            return self.value
        return self.to_celsius() * 9/5 + 32


    def to_kelvin(self) -> float:
        """Return the temperature in Kelvin."""
        if self.unit == "K":
            return self.value
        if self.value == self._ABSOLUTE_ZERO[self.unit]:
            return 0.0
        return self.to_celsius() + 273.15


    def convert_to(self, target_unit: str) -> "Temperature":
        """Return a new Temperature expressed in the requested unit."""
        unit = self._validate_unit(target_unit)
        if unit == "C":
            value = self.to_celsius()
        elif unit == "F":
            value = self.to_fahrenheit()
        else:
            value = self.to_kelvin()

        # At absolute zero, use the exact boundary as the target unit.
        if self.value == self._ABSOLUTE_ZERO[self.unit]:
            value = self._ABSOLUTE_ZERO[unit]
        return Temperature(value, unit)


    def __repr__(self) -> str:
        """Return a string representation of the temperature."""
        return f"Temperature({self.value!r}, {self.unit!r})"


    def __str__(self) -> str:
        """Return the temperature formatted to two decimal places."""
        symbol = "K" if self.unit == "K" else f"°{self.unit}"
        return f"{self.value:.2f} {symbol}"


    def __eq__(self, other: object) -> bool:
        """Return whether temperatures agree with the absolute tolerance."""
        if not isinstance(other, Temperature):
            return NotImplemented
        return isclose(
            self.to_kelvin(),
            other.to_kelvin(),
            rel_tol=0.0,
            abs_tol=self._TOLERANCE,
        )


    def __lt__(self, other: object) -> bool:
        """Return whether the temperature is colder by more than the tolerance."""
        if not isinstance(other, Temperature):
            return NotImplemented
        return self != other and self.to_kelvin() < other.to_kelvin()
```

Created a temperature class which represents a temperature in 'K', 'F' and 'C'. The class will work with lowercase symbols like 'c' by turning them into uppercase letters. Raises ValueError if value is below absolute zero.
Methods like 'to_fahrenheit' returns a number, while 'convert_to' creates a new temperature object. To compare temperatures, the class converts both values to Kelvin.

Task 2: Fractions.

```python
%%writefile fraction.py
from functools import total_ordering
from math import gcd

@total_ordering
class Fraction:
    """Represent a fraction in its simplest form with a positive denominator."""
    

    def __init__(self, numerator: int, denominator: int) -> None:
        """Create  a fraction from to integers with positive denominator."""
        if type(numerator) is not int or type(denominator) is not int:
            raise TypeError("The numerator and denominator must be integers,")
        if denominator <= 0:
            raise ValueError("The denominator must be greater than zero.")

        divisor = gcd(numerator, denominator)
        self._numerator = numerator // divisor
        self._denominator = denominator // divisor
        

    @property
    def numerator(self) -> int:
        """Return the numerator of the simplified fraction."""
        return self._numerator


    @property
    def denominator(self) -> int:
        """Return the denominator of the simplified fraction."""
        return self._denominator


    def __eq__(self, other: object) -> bool:
        """Return whether two fractions have the same value."""
        if not isinstance(other, Fraction):
            return NotImplemented
        return (
            self.numerator == other.numerator
            and self.denominator == other.denominator
        )
    

    def __lt__(self, other: object) -> bool:
        """Return whether this fraction is smaller than the other fraction."""
        if not isinstance(other, Fraction):
            return NotImplemented
        return self.numerator * other.denominator < other.numerator * self.denominator
    

    def __add__(self, other: "Fraction") -> "Fraction":
        """Return the sum of two fractions."""
        if not isinstance(other, Fraction):
            return NotImplemented

        numerator = (
            self.numerator * other.denominator
            + other.numerator * self.denominator
        )
        denominator = self.denominator * other.denominator
        return Fraction(numerator, denominator)


    def __mul__(self, other: "Fraction") -> "Fraction":
        """Return the product of two fractions."""
        if not isinstance(other, Fraction):
            return NotImplemented

        numerator = self.numerator * other.numerator
        denominator = self.denominator * other.denominator
        return Fraction(numerator, denominator)


    def __str__(self) -> str:
        """Return the fraction in numerator/denominator form."""
        return f"{self.numerator}/{self.denominator}"


    def __repr__ (self) -> str:
        """Return a representation showing how to create the fraction."""
        return f"Fraction({self.numerator}, {self.denominator})"

```

The class accepts an integer numerator and a positive integer denominator. It uses the greatest common divisor to simplify the fractions. Fractions are compared using integers. Addition and multiplication return a new, simplified fraction.
